## Preprocess Policy Document

### Subtask:
Process the loaded policy document by splitting it into manageable chunks and generating embeddings for each chunk to facilitate question answering.

```python
import nltk
from sentence_transformers import SentenceTransformer

nltk.download('punkt')

print("NLTK and SentenceTransformer imported and punkt tokenizer downloaded successfully.")
```

```python
import nltk
nltk.download('punkt')
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
sentences = nltk.sent_tokenize(policy_document_content)
sentence_embeddings = model.encode(sentences)

# Store sentences and embeddings
processed_policy = []
for i, sentence in enumerate(sentences):
    processed_policy.append({
        'sentence': sentence,
        'embedding': sentence_embeddings[i]
    })

print(f"Policy document processed into {len(sentences)} sentences.")
print(f"First sentence: {processed_policy[0]['sentence']}")
print(f"Shape of embeddings for the first sentence: {processed_policy[0]['embedding'].shape}")
```

## Build Question Answering System

### Subtask:
Develop the core QA logic. This involves taking a user's question, generating its embedding, finding the most relevant policy document chunks based on similarity, and then using a suitable language model to extract or generate an answer from the retrieved context.


**Reasoning**:
I will define the `answer_question` function as specified in the instructions, which includes generating embeddings for the question, calculating cosine similarity with processed policy sentences, retrieving the top N most similar sentences, and constructing an answer from them.



```python
from sentence_transformers import util

def answer_question(question, top_n=3):
    # Generate embedding for the input question
    question_embedding = model.encode(question)

    # Calculate cosine similarity between question embedding and all sentence embeddings
    # Extract embeddings from processed_policy for similarity calculation
    all_sentence_embeddings = [item['embedding'] for item in processed_policy]
    cosine_scores = util.cos_sim(question_embedding, all_sentence_embeddings)[0]

    # Get the top_n most similar sentences
    top_results = []
    for i in range(top_n):
        # Get the index of the highest score that hasn't been added yet
        best_idx = cosine_scores.argmax()
        if cosine_scores[best_idx] > 0: # Only consider scores greater than 0
            top_results.append({
                'sentence': processed_policy[best_idx]['sentence'],
                'score': cosine_scores[best_idx].item() # Convert tensor to float
            })
            # Set the score to a very low number so it's not chosen again
            cosine_scores[best_idx] = -1.0
        else:
            break # No more relevant sentences found

    # Construct the answer
    if top_results:
        answer = "Based on the policy: "
        for res in top_results:
            answer += res['sentence'] + " "
        return answer.strip()
    else:
        return "I'm sorry, I couldn't find relevant information in the policy document for your question."

print("answer_question function defined successfully.")
```

## Create Interactive Chatbot Interface

### Subtask:
Implement a simple interactive command-line or web-based interface (e.g., using `gradio` or `streamlit`) that allows users to type questions and receive answers from the QA system.


**Reasoning**:
I will import the `gradio` library and then define and launch a `gr.Interface` using the previously defined `answer_question` function, setting up appropriate input and output components, and adding a title and description as instructed.



```python
import gradio as gr

# Create the Gradio interface
iface = gr.Interface(
    fn=answer_question,
    inputs=gr.Textbox(lines=2, placeholder="Ask a question about the policy..."),
    outputs=gr.Textbox(),
    title="Company Policy Chatbot",
    description="Ask any question about the company policy document and get relevant answers."
)

# Launch the interface
iface.launch(debug=True)

print("Gradio interface defined and launched successfully.")
```

**Reasoning**:
The `UserWarning` from `sentence_transformers` indicates that creating a tensor from a list of numpy arrays is slow. To fix this and improve performance, the `answer_question` function should convert the list of sentence embeddings to a single numpy array using `np.array()` before computing cosine similarity. This change will be applied to the `answer_question` function in cell `db1601b8`.



```python
from sentence_transformers import util
import numpy as np # Import numpy

def answer_question(question, top_n=3):
    # Generate embedding for the input question
    question_embedding = model.encode(question)

    # Calculate cosine similarity between question embedding and all sentence embeddings
    # Extract embeddings from processed_policy for similarity calculation
    all_sentence_embeddings = [item['embedding'] for item in processed_policy]

    # Convert list of numpy arrays to a single numpy array to avoid UserWarning
    all_sentence_embeddings_np = np.array(all_sentence_embeddings)

    cosine_scores = util.cos_sim(question_embedding, all_sentence_embeddings_np)[0]

    # Get the top_n most similar sentences
    top_results = []
    for i in range(top_n):
        # Get the index of the highest score that hasn't been added yet
        best_idx = cosine_scores.argmax()
        if cosine_scores[best_idx] > 0: # Only consider scores greater than 0
            top_results.append({
                'sentence': processed_policy[best_idx]['sentence'],
                'score': cosine_scores[best_idx].item() # Convert tensor to float
            })
            # Set the score to a very low number so it's not chosen again
            cosine_scores[best_idx] = -1.0
        else:
            break # No more relevant sentences found

    # Construct the answer
    if top_results:
        answer = "Based on the policy: "
        for res in top_results:
            answer += res['sentence'] + " "
        return answer.strip()
    else:
        return "I'm sorry, I couldn't find relevant information in the policy document for your question."

print("answer_question function defined successfully with performance improvement.")
```
